<h2>Formating of DB-Data<h2>

```python
from hilfsfunktionen.formating import get_unchanged_table, get_formated_table, get_table_with_keys, create_n_sentences
df_collection = {}
df_collection["unchanged_table"] = get_unchanged_table()
df_collection["formated_table"] = get_formated_table()
df_collection["formated_table_with_keys"] = get_table_with_keys(df_collection["formated_table"])
df_collection["n_sentences"] = {}
for n in [1, 5, 20, 50, 100]:
    df_collection["n_sentences"][f"{n}"] = create_n_sentences(df_collection["formated_table_with_keys"], n)
```

<h2>Model Creation<h2>

```python
from hilfsfunktionen.model_training import ModelGenerator
from pathlib import Path

base_params = {
            "vector_size": 50,
            "window": 4,
            "epochs": 30,
            "min_count": 1,
            "negative": 5,
            "sample": 1e-3,
            "hs": 0,
            "alpha": 0.025,
            "seed": 1,
            "num_sentences": 10000
        }
model_gen = ModelGenerator()
model_gen.set_base_values(base_params)
models, categories = model_gen.create_many_models(vector_sizes=[50],models = {}, n_sentences_tokenized = df_collection["n_sentences"], sentences_tokenized = df_collection["n_sentences"]["20"],base_dir = Path.cwd(), verbose = False)
```

<h2>Durchschnittliche Satzvektoren berechnen<h2>

```python
from hilfsfunktionen.sentence_vectors import create_average_sentence_vectors
average_sentence_vectors = {}
for category in categories:
    print(f"(-------------Vector calculation of category: {category.upper()}-------------)")
    average_sentence_vectors[category] = []
    for model in models[category]:
        average_sentence_vectors[category].append([create_average_sentence_vectors(model, df_collection["formated_table_with_keys"], base_dir = Path.cwd(), verbose=False),model[1]])
    print(f"(-------------Finished Vector calculation of category: {category.upper()}-------------)")
```

<h2>Vektoren-Test Textbasiert<h2>

```python
from hilfsfunktionen.collect_test_data import get_all_results
    
all_model_results = get_all_results(models, categories, df_with_keys=df_collection["formated_table_with_keys"])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# ============================================================================
# 1. GENERISCHE LINE PLOT FUNKTION FÜR PARAMETER-ENTWICKLUNG
# ============================================================================

def plot_parameter_development(all_model_results, category, parameter_key, 
                               metric='both', figsize=(14, 6)):
    """
    Erstellt Line Plots für die Entwicklung von Qualitätsmetriken über einen Parameter.
    
    Parameters:
    -----------
    all_model_results : list
        Liste mit allen Testergebnissen
    category : str
        Test_Category (z.B. 'vector', 'windowsize', 'epoch', 'sentences')
    parameter_key : str
        Schlüssel des Parameters in params dict
        (z.B. 'Vector Size', 'Window', 'Epochs', 'Satzanzahl')
    metric : str
        'semantic', 'neighbor' oder 'both'
    figsize : tuple
        Größe der Figur
    """
    
    # Daten filtern
    filtered_results = [r for r in all_model_results 
                       if r['params']['Test_Category'] == category]
    
    if not filtered_results:
        print(f"Keine Daten für Kategorie '{category}' gefunden.")
        return
    
    # Algorithmen identifizieren (CBOW vs SGNS)
    algorithms = list(set([r['params']['Algorithmus'] for r in filtered_results]))
    
    # Subplots erstellen
    n_plots = 2 if metric == 'both' else 1
    fig, axes = plt.subplots(1, n_plots, figsize=figsize)
    if n_plots == 1:
        axes = [axes]
    
    # Farbpalette
    colors = {'CBOW': '#3498db', 'SGNS': '#e74c3c'}
    markers = {'CBOW': 'o', 'SGNS': 's'}
    
    # ========================================================================
    # PLOT 1: SEMANTIC SIMILARITY
    # ========================================================================
    if metric in ['semantic', 'both']:
        ax = axes[0] if metric == 'both' else axes[0]
        
        # Daten pro Token-Paar sammeln
        token_pairs = []
        for result in filtered_results:
            for sim_test in result['basic']['semantic_similarity']:
                if sim_test['pair'] not in token_pairs:
                    token_pairs.append(sim_test['pair'])
        
        # Für jedes Token-Paar eine Linie pro Algorithmus
        for pair in token_pairs:
            for alg in algorithms:
                param_values = []
                similarities = []
                
                for result in filtered_results:
                    if result['params']['Algorithmus'] == alg:
                        param_val = result['params'][parameter_key]
                        
                        # Similarity für dieses Paar finden
                        for sim_test in result['basic']['semantic_similarity']:
                            if sim_test['pair'] == pair and sim_test['similarity'] is not None:
                                param_values.append(param_val)
                                similarities.append(sim_test['similarity'])
                                break
                
                if param_values:
                    # Sortieren nach Parameter-Wert
                    sorted_data = sorted(zip(param_values, similarities))
                    param_values, similarities = zip(*sorted_data)
                    
                    # Plot
                    ax.plot(param_values, similarities, 
                           marker=markers[alg], 
                           color=colors[alg],
                           alpha=0.6,
                           linewidth=2,
                           markersize=6,
                           label=f"{pair} ({alg})")
        
        ax.set_xlabel(parameter_key, fontsize=12, fontweight='bold')
        ax.set_ylabel('Cosine Similarity', fontsize=12, fontweight='bold')
        ax.set_title(f'Semantische Ähnlichkeit über {parameter_key}', 
                    fontsize=13, fontweight='bold', pad=15)
        ax.grid(True, alpha=0.3, linestyle='--')
        ax.legend(loc='best', fontsize=8, ncol=2)
        ax.axhline(y=0.5, color='gray', linestyle=':', alpha=0.5)
    
    # ========================================================================
    # PLOT 2: NEIGHBOR QUALITY
    # ========================================================================
    if metric in ['neighbor', 'both']:
        ax = axes[1] if metric == 'both' else axes[0]
        
        # Durchschnittliche Scores pro Kategorie berechnen
        token_categories = []
        for result in filtered_results:
            for neighbor_test in result['basic']['neighbors']:
                if neighbor_test['category'] not in token_categories:
                    token_categories.append(neighbor_test['category'])
        
        # Für jede Token-Kategorie eine Linie pro Algorithmus
        for token_cat in token_categories:
            for alg in algorithms:
                param_values = []
                avg_scores = []
                
                for result in filtered_results:
                    if result['params']['Algorithmus'] == alg:
                        param_val = result['params'][parameter_key]
                        
                        # Scores für diese Kategorie sammeln
                        scores = [n['score'] for n in result['basic']['neighbors'] 
                                 if n['category'] == token_cat]
                        
                        if scores:
                            param_values.append(param_val)
                            avg_scores.append(np.mean(scores))
                
                if param_values:
                    # Sortieren
                    sorted_data = sorted(zip(param_values, avg_scores))
                    param_values, avg_scores = zip(*sorted_data)
                    
                    # Plot
                    ax.plot(param_values, avg_scores,
                           marker=markers[alg],
                           color=colors[alg],
                           alpha=0.7,
                           linewidth=2.5,
                           markersize=7,
                           label=f"{token_cat} ({alg})")
        
        ax.set_xlabel(parameter_key, fontsize=12, fontweight='bold')
        ax.set_ylabel('Average Neighbor Quality Score', fontsize=12, fontweight='bold')
        ax.set_title(f'Nachbar-Qualität über {parameter_key}', 
                    fontsize=13, fontweight='bold', pad=15)
        ax.set_ylim(0, 1.1)
        ax.grid(True, alpha=0.3, linestyle='--')
        ax.legend(loc='best', fontsize=9)
        ax.axhline(y=1.0, color='red', linestyle='--', alpha=0.5, linewidth=1.5)
    
    plt.tight_layout()
    return fig


# ============================================================================
# 2. SPEZIELLE WRAPPER-FUNKTIONEN FÜR JEDE KATEGORIE
# ============================================================================

def plot_vector_size_development(all_model_results, metric='both', figsize=(14, 6)):
    """
    Line Plot für Vector Size Entwicklung.
    """
    return plot_parameter_development(
        all_model_results, 
        category='vector',
        parameter_key='Vector Size',
        metric=metric,
        figsize=figsize
    )


def plot_window_size_development(all_model_results, metric='both', figsize=(14, 6)):
    """
    Line Plot für Window Size Entwicklung.
    """
    return plot_parameter_development(
        all_model_results,
        category='windowsize',
        parameter_key='Window',
        metric=metric,
        figsize=figsize
    )


def plot_epoch_development(all_model_results, metric='both', figsize=(14, 6)):
    """
    Line Plot für Epochs Entwicklung.
    """
    return plot_parameter_development(
        all_model_results,
        category='epoch',
        parameter_key='Epochs',
        metric=metric,
        figsize=figsize
    )


def plot_sentence_count_development(all_model_results, metric='both', figsize=(14, 6)):
    """
    Line Plot für Sentence Count Entwicklung.
    """
    return plot_parameter_development(
        all_model_results,
        category='sentences',
        parameter_key='Satzanzahl',
        metric=metric,
        figsize=figsize
    )


# ============================================================================
# 3. KOMBINIERTE 4-IN-1 ÜBERSICHT
# ============================================================================

def plot_all_parameter_developments(all_model_results, metric='neighbor', figsize=(16, 12)):
    """
    Erstellt eine 2x2 Grid-Übersicht über alle 4 Hauptparameter.
    
    Parameters:
    -----------
    all_model_results : list
        Liste mit allen Testergebnissen
    metric : str
        'semantic' oder 'neighbor' (nicht 'both', da sonst zu voll)
    figsize : tuple
        Größe der Figur
    """
    
    fig, axes = plt.subplots(2, 2, figsize=figsize)
    
    configs = [
        ('vector', 'Vector Size', axes[0, 0]),
        ('windowsize', 'Window', axes[0, 1]),
        ('epoch', 'Epochs', axes[1, 0]),
        ('sentences', 'Satzanzahl', axes[1, 1])
    ]
    
    colors = {'CBOW': '#3498db', 'SGNS': '#e74c3c'}
    markers = {'CBOW': 'o', 'SGNS': 's'}
    
    for category, param_key, ax in configs:
        filtered_results = [r for r in all_model_results 
                           if r['params']['Test_Category'] == category]
        
        if not filtered_results:
            ax.text(0.5, 0.5, f'Keine Daten für {category}', 
                   ha='center', va='center', transform=ax.transAxes)
            continue
        
        algorithms = list(set([r['params']['Algorithmus'] for r in filtered_results]))
        
        if metric == 'semantic':
            # Token-Paare
            token_pairs = []
            for result in filtered_results:
                for sim_test in result['basic']['semantic_similarity']:
                    if sim_test['pair'] not in token_pairs:
                        token_pairs.append(sim_test['pair'])
            
            for pair in token_pairs:
                for alg in algorithms:
                    param_values = []
                    similarities = []
                    
                    for result in filtered_results:
                        if result['params']['Algorithmus'] == alg:
                            param_val = result['params'][param_key]
                            for sim_test in result['basic']['semantic_similarity']:
                                if sim_test['pair'] == pair and sim_test['similarity'] is not None:
                                    param_values.append(param_val)
                                    similarities.append(sim_test['similarity'])
                                    break
                    
                    if param_values:
                        sorted_data = sorted(zip(param_values, similarities))
                        param_values, similarities = zip(*sorted_data)
                        ax.plot(param_values, similarities,
                               marker=markers[alg], color=colors[alg],
                               alpha=0.5, linewidth=1.5, markersize=4)
            
            ax.set_ylabel('Similarity', fontsize=10)
            ax.axhline(y=0.5, color='gray', linestyle=':', alpha=0.5)
        
        elif metric == 'neighbor':
            # Durchschnittlicher Score über alle Token-Kategorien
            for alg in algorithms:
                param_values = []
                overall_avg_scores = []
                
                for result in filtered_results:
                    if result['params']['Algorithmus'] == alg:
                        param_val = result['params'][param_key]
                        all_scores = [n['score'] for n in result['basic']['neighbors']]
                        
                        if all_scores:
                            param_values.append(param_val)
                            overall_avg_scores.append(np.mean(all_scores))
                
                if param_values:
                    sorted_data = sorted(zip(param_values, overall_avg_scores))
                    param_values, overall_avg_scores = zip(*sorted_data)
                    
                    ax.plot(param_values, overall_avg_scores,
                           marker=markers[alg], color=colors[alg],
                           alpha=0.8, linewidth=2.5, markersize=6,
                           label=alg)
            
            ax.set_ylabel('Avg Quality Score', fontsize=10)
            ax.set_ylim(0, 1.1)
            ax.axhline(y=1.0, color='red', linestyle='--', alpha=0.5)
            ax.legend(loc='best', fontsize=8)
        
        ax.set_xlabel(param_key, fontsize=10, fontweight='bold')
        ax.set_title(f'{param_key} Entwicklung', fontsize=11, fontweight='bold')
        ax.grid(True, alpha=0.3, linestyle='--')
    
    metric_title = 'Semantische Ähnlichkeit' if metric == 'semantic' else 'Nachbar-Qualität'
    fig.suptitle(f'Parameter-Entwicklung: {metric_title}', 
                fontsize=15, fontweight='bold', y=0.995)
    
    plt.tight_layout()
    return fig

```

```python

fig1 = plot_vector_size_development(all_model_results, metric='both')
plt.show()

fig2 = plot_window_size_development(all_model_results, metric='neighbor')
plt.show()

fig3 = plot_epoch_development(all_model_results, metric='semantic')
plt.show()

fig4 = plot_sentence_count_development(all_model_results, metric='both')
plt.show()

fig5 = plot_all_parameter_developments(all_model_results, metric='neighbor')
plt.show()
```
